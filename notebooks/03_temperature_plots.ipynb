# 03 - Temperature plots

Plots each city's average monthly temperature for the cards. Output goes to `outputs/temp_plots/` (not committed).

```python
import os
import pandas as pd
import math
import matplotlib.pyplot as plt

os.makedirs('../outputs/temp_plots', exist_ok=True)

# Load the CSV file
csv_path = '../data/processed/city_list_set_1.csv'
df = pd.read_csv(csv_path)

# Define the month columns
month_cols = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']
x_ticks_labels = ['Jan', 'Mar', 'Jun', 'Sep', 'Dec']
x_ticks_locs = [0, 2, 5, 8, 11]

# Get the overall min and max temperature values
min_temp = df[month_cols].min().min()
max_temp = df[month_cols].max().max()

# Get round numbers for figure bounds
# lower_lim = math.floor(min_temp/10)*10
lower_lim = 0
# upper_lim = math.ceil(max_temp/10)*10
upper_lim = 100

# Iterate through each city
for index, row in df.iterrows():
    city_name = row['City']
    state_id = row['State ID']
    temps = row[month_cols].values

    # Create the line plot
    fontsize = 30
    plt.figure(figsize=(10, 6), facecolor='none')  # Adjust figure size as needed
    ax = plt.gca()
    ax.set_facecolor((1, 1, 1, 0.25))
    ax.plot(month_cols, temps, marker='o', linestyle='-', color='black', linewidth=5, markersize=20)
    # plt.title(f'Average Monthly Temperatures for {city_name}, {state_id}')
    # plt.xlabel('Month')
    plt.ylabel('Avg Temp ($\\circ$F)', fontsize=fontsize)
    plt.ylim(lower_lim, upper_lim)  # Set y-axis bounds with a small buffer
    plt.yticks([0, 25, 50, 75, 100], fontsize=fontsize)
    plt.grid(False)
    plt.xticks(rotation=45, fontsize=fontsize)
    # plt.xticks(x_ticks_locs, x_ticks_labels, rotation=45)  # Rotate x-axis labels for better readability
    plt.tick_params(axis='both', which='both', length=0) 
    plt.tight_layout()

    # Increase the thickness of the plot borders (spines)
    for spine in ax.spines.values():
        spine.set_linewidth(4)  # You can adjust this value

    # Define the export file name and path
    filename = f"{state_id}_{city_name}.png"
    export_path = f"../outputs/temp_plots/{filename}"

    # Export the plot as a PNG file
    plt.savefig(export_path, dpi=600)
    plt.close()  # Close the figure to free up memory

print("Temperature plots have been created and saved.")
```

```python
import os
from PIL import Image

def resize_png_files(folder_path, target_width, target_height):
    """
    Resizes all .png files in the specified folder to the target dimensions.

    Args:
        folder_path (str): The path to the folder containing the .png files.
        target_width (int): The desired width of the resized images.
        target_height (int): The desired height of the resized images.
    """
    try:
        if not os.path.isdir(folder_path):
            print(f"Error: Folder not found at '{folder_path}'")
            return

        for filename in os.listdir(folder_path):
            if filename.lower().endswith(".png"):
                file_path = os.path.join(folder_path, filename)
                try:
                    img = Image.open(file_path)
                    resized_img = img.resize((target_width, target_height))
                    resized_img.save(file_path)  # Overwrites the original file
                    print(f"Resized: {filename} to {target_width}x{target_height}")
                except Exception as e:
                    print(f"Error processing {filename}: {e}")

        print("Resizing complete.")

    except Exception as e:
        print(f"An unexpected error occurred: {e}")

if __name__ == "__main__":
    folder_path = '../outputs/temp_plots'
    target_width = 250
    target_height = 150

    resize_png_files(folder_path, target_width, target_height)
```
